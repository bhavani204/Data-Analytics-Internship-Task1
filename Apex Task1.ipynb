```python
#Reading Excel files using Pandas
import pandas as pd
df = pd.read_excel("ApexPlanet_DataAnalytics_Dataset.xlsx")
df.info()
df.head(20)
```

```python
#statistical data
df.describe(include='all')
```

```python
#checking null values in each column
print(df.isnull().sum())
```

```python
#Filling null values using mean,mode,median
df['Age']=df['Age'].fillna(df['Age'].mean())
df['City']=df['City'].fillna(df['City'].mode()[0])

```

```python
#verifying null value exists or not
df.isnull().sum()
```

```python
#Checking the duplicated records
df.duplicated().sum()
```

```python
#Removing duplicate records
df=df.drop_duplicates()
```

```python
#verifying the records
df.duplicated().sum()
```

```python
#Outlier Detection
Q1=df['Total_Sales'].quantile(0.25)
Q3=df['Total_Sales'].quantile(0.75)
IQR=Q3-Q1
lower_fence=Q1-1.5*IQR
upper_fence=Q3+1.5*IQR
print(lower_fence)
print(upper_fence)
```

```python
#filter the data
filter_data=df[df['Total_Sales'].between(lower_fence,upper_fence)]
print(filter_data)
```

```python
df.to_excel("ApexPlanet_DataAnalytics_Dataset.xlsx", index=False)
```
